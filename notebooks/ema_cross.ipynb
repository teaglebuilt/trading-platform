```python
from __future__ import annotations
from typing import List
from strategy_engine.interfaces import Strategy, Signal, OrderIntent, Venue


class EmaCross(Strategy):
    name = "ema_cross"

    def evaluate(self, sig: Signal) -> List[OrderIntent]:
        """
        Example of a local strategy that trusts external signal fields.
        In reality, you'd compute EMAs from market data pulled here or precomputed in orchestrator.
        """
        side = sig.side
        qty = float(sig.meta.get("qty", 1))
        # choose venue by symbol class
        if sig.symbol.endswith("USD"):
            venue = getattr(Venue, "ALPACA_CRYPTO", "alpaca_crypto")
        else:
            venue = getattr(Venue, "ALPACA_STOCKS", "alpaca_stocks")
        return [OrderIntent(venue=venue, symbol=sig.symbol, side=side, qty=qty, type="market", tag=self.name)]
```
